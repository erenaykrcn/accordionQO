```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

def phase_wrap(x):
    return (x + np.pi) % (2 * np.pi) - np.pi

def detect_vortices_from_phase(phase):
    th00 = phase
    th10 = np.roll(phase, shift=-1, axis=0)
    th11 = np.roll(np.roll(phase, shift=-1, axis=0), shift=-1, axis=1)
    th01 = np.roll(phase, shift=-1, axis=1)

    d1 = phase_wrap(th10 - th00)
    d2 = phase_wrap(th11 - th10)
    d3 = phase_wrap(th01 - th11)
    d4 = phase_wrap(th00 - th01)

    winding = (d1 + d2 + d3 + d4) / (2 * np.pi)
    winding_int = np.rint(winding).astype(int)

    vort = np.argwhere(winding_int == 1)
    antiv = np.argwhere(winding_int == -1)
    return vort, antiv
```

```python
def detect_vortices_from_phase(phase, density, density_threshold=0.07):
    """
    phase: 2D numpy array (angle field)
    density: 2D numpy array (|psi|^2)
    density_threshold: fraction of max density

    Returns:
        vortices, antivortices (arrays of indices)
    """

    # --- normalize threshold relative to peak density ---
    nmax = density.max()
    mask = density > (density_threshold * nmax)

    # --- plaquette corners ---
    th00 = phase
    th10 = np.roll(phase, shift=-1, axis=0)
    th11 = np.roll(np.roll(phase, shift=-1, axis=0), shift=-1, axis=1)
    th01 = np.roll(phase, shift=-1, axis=1)

    # --- density mask on corners ---
    m00 = mask
    m10 = np.roll(mask, shift=-1, axis=0)
    m11 = np.roll(np.roll(mask, shift=-1, axis=0), shift=-1, axis=1)
    m01 = np.roll(mask, shift=-1, axis=1)

    # Only keep plaquettes where ALL 4 corners are dense
    valid = m00 & m10 & m11 & m01

    # --- phase differences ---
    d1 = phase_wrap(th10 - th00)
    d2 = phase_wrap(th11 - th10)
    d3 = phase_wrap(th01 - th11)
    d4 = phase_wrap(th00 - th01)

    winding = (d1 + d2 + d3 + d4) / (2 * np.pi)
    winding_int = np.rint(winding).astype(int)

    # --- apply mask ---
    winding_int[~valid] = 0

    vort = np.argwhere(winding_int == 1)
    antiv = np.argwhere(winding_int == -1)

    return vort, antiv
```

```python
def update(frame):
    ph1 = phase1[frame]
    ph2 = phase2[frame]
    d1 = density1[frame]
    d2 = density2[frame]

    thresh1 = 1e-5 * d1.max()
    thresh2 = 1e-5 * d2.max()
    # --- mask threshold ---
    """
    mask1 = d1 > thresh1
    mask2 = d2 > thresh2
    """
    # --- circular mask centered at (50,50), diameter 40 ---
    cx, cy = 50, 50
    radius = 20
    Y, X = np.ogrid[:d1.shape[0], :d1.shape[1]]
    circle_mask = (X - cx)**2 + (Y - cy)**2 <= radius**2
    mask1 = (d1 > thresh1) & circle_mask
    mask2 = (d2 > thresh2) & circle_mask

    ph1_plot = np.where(mask1, ph1, np.nan)
    ph2_plot = np.where(mask2, ph2, np.nan)
    

    # --- vortex detection (masked version!) ---
    vort1, antiv1 = detect_vortices_from_phase(ph1, d1)
    vort2, antiv2 = detect_vortices_from_phase(ph2, d2)

    # --- update images ---
    im1.set_data(ph1_plot.T)
    im2.set_data(ph2_plot.T)

    # --- scatter updates ---
    sc_v1.set_offsets(np.c_[vort1[:, 1], vort1[:, 0]] if len(vort1) else np.empty((0, 2)))
    sc_av1.set_offsets(np.c_[antiv1[:, 1], antiv1[:, 0]] if len(antiv1) else np.empty((0, 2)))

    sc_v2.set_offsets(np.c_[vort2[:, 1], vort2[:, 0]] if len(vort2) else np.empty((0, 2)))
    sc_av2.set_offsets(np.c_[antiv2[:, 1], antiv2[:, 0]] if len(antiv2) else np.empty((0, 2)))

    fig.suptitle(f"Frame {frame}")
    return im1, im2, sc_v1, sc_av1, sc_v2, sc_av2


T, J = 120, 0
# Load data
data = np.load("p/"+f"out_T_{T:.3f}_J_{J:.3f}".replace(".", "p")+"/frames.npz") 
phase1 = data["frames_phase_1"]
phase2 = data["frames_phase_2"]
density1 = data["frames_density_1"]
density2 = data["frames_density_2"]

# Setup figure
fig, axes = plt.subplots(1, 2, figsize=(12, 6))
# Layer 1
im1 = axes[0].imshow(phase1[0].T, origin="lower", cmap="twilight", vmin=-np.pi, vmax=np.pi)
sc_v1 = axes[0].scatter([], [], s=30, marker="o", label="vortex", color='red')
sc_av1 = axes[0].scatter([], [], s=30, marker="x", label="antivortex", color='yellow')
axes[0].set_title("Layer 1")
axes[0].legend()
# Layer 2
im2 = axes[1].imshow(phase2[0].T, origin="lower", cmap="twilight", vmin=-np.pi, vmax=np.pi)
sc_v2 = axes[1].scatter([], [], s=30, marker="o", label="vortex", color='red')
sc_av2 = axes[1].scatter([], [], s=30, marker="x", label="antivortex", color='yellow')
axes[1].set_title("Layer 2")
axes[1].legend()
ani = FuncAnimation(fig, update, frames=len(phase1), interval=100, blit=False)
ani.save(f"gifsv2/BILAYER_T_{T:.3f}_J_{J:.3f}.gif", writer="ffmpeg", fps=5, dpi=100)
plt.show()
```
